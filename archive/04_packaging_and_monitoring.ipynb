# Notebook 04 - Packaging and monitoring (Step 11)

**Goal:** show that the model saved by notebook 03 can be reloaded and used on its own, and decide how we would watch it after deployment.

1. Reload the saved pipeline and **prove** it gives the same answers as before.
2. Score one new application.
3. Watch for **drift**: do new loans look different from training loans, and does accuracy fall over time?
4. Write a **retraining rule** and a **model card**.

```python
from pathlib import Path
import json, platform
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib, sklearn
from sklearn.metrics import roc_auc_score

ROOT = Path("..")
ARTIFACTS = ROOT / "artifacts"
FIG_DIR = ARTIFACTS / "figures"
sns.set_theme(style="whitegrid")
pd.set_option("display.float_format", "{:,.3f}".format)
pd.set_option("display.width", 140)

def save_fig(name):
    plt.savefig(FIG_DIR / f"{name}.png", dpi=150, bbox_inches="tight")

data = pd.read_parquet(ROOT / "data/processed/clean.parquet")
features = json.loads((ROOT / "data/processed/features.json").read_text())
NUMERIC, CATEGORICAL = features["numeric"], features["categorical"]
FEATURE_COLS = NUMERIC + CATEGORICAL
train = data[data["split"] == "train"]
test = data[data["split"] == "test"].sort_values("issue_d").reset_index(drop=True)

model = joblib.load(ARTIFACTS / "model.joblib")
config = json.loads((ARTIFACTS / "threshold.json").read_text())
metrics = pd.read_csv(ARTIFACTS / "metrics.csv", index_col=0)
print("Loaded model from disk:", config["model"], "| threshold", round(config["threshold"], 3))
```

## Step 11 - Package, deploy, monitor

### 11.1 Round-trip check
The model was saved in notebook 03 and has just been loaded from disk. If saving and loading worked, scoring the test set now must give **exactly** the ROC-AUC that notebook 03 recorded in `metrics.csv`.

```python
test_scores = model.predict_proba(test[FEATURE_COLS])[:, 1]
reloaded_auc = roc_auc_score(test["default"], test_scores)
saved_auc = metrics.loc[config["model"], "ROC-AUC"]

print(f"ROC-AUC after reloading: {reloaded_auc:.6f}")
print(f"ROC-AUC saved in notebook 03: {saved_auc:.6f}")
assert abs(reloaded_auc - saved_auc) < 1e-6, "The reloaded model gives different answers!"
print("Round trip OK: identical.")
```

### 11.2 Score one new application
A real service receives one application at a time. The pipeline accepts a raw row (a one-row table) and does all the imputing, scaling and encoding itself. The application below is **invented for the demonstration** and is not a real loan. We also check the pipeline survives a missing value and a category it has never seen.

```python
new_application = pd.DataFrame([{
    "loan_amnt": 12000, "int_rate": 13.5, "installment": 408.0, "term_months": 36, "emp_length_years": 4,
    "dti": 18.0, "delinq_2yrs": 0, "fico_avg": 690, "inq_last_6mths": 1, "open_acc": 10, "pub_rec": 0,
    "revol_util": 55, "total_acc": 22, "mort_acc": 1, "pub_rec_bankruptcies": 0, "credit_history_years": 14,
    "log_annual_inc": np.log1p(65000), "log_revol_bal": np.log1p(11000), "loan_to_income": 12000 / 65000,
    "grade": "C", "home_ownership": "RENT", "verification_status": "Verified",
    "purpose": "debt_consolidation", "application_type": "Individual"}])

def decide(application):
    score = model.predict_proba(application[FEATURE_COLS])[:, 1][0]
    return score, "FLAG as high risk" if score >= config["threshold"] else "pass"

score, decision = decide(new_application)
print(f"Predicted default probability: {score:.3f} -> {decision} (threshold {config['threshold']:.3f})")

# Robustness: a missing value and an unseen category must not crash the pipeline
messy = new_application.copy()
messy.loc[0, "mort_acc"] = np.nan
messy.loc[0, "purpose"] = "space_travel"
score2, decision2 = decide(messy)
print(f"With a missing mort_acc and an unseen purpose: {score2:.3f} -> {decision2} (no crash)")
```

### 11.3 Monitoring 1 - do new loans look different from training loans?
For each numeric feature we measure how far the **test-period median** has moved from the **training median**, in training standard deviations: `(median_test - median_train) / std_train`. We use the median because a few extreme values can drag a mean a long way (we show below that this really happens). A second check counts test rows that fall **outside the range seen in training**, where the model has never been shown any example.

**Figure 1.** Standardised shift in the median of each numeric feature, test period vs training period (top 10 by size).

```python
shift = ((test[NUMERIC].median() - train[NUMERIC].median()) / train[NUMERIC].std()).sort_values(key=abs, ascending=False)
mean_shift = (test[NUMERIC].mean() - train[NUMERIC].mean()) / train[NUMERIC].std()
top = shift.head(10).iloc[::-1]

plt.figure(figsize=(7, 4.5))
plt.barh(top.index, top.values, color=["tab:red" if v > 0 else "tab:blue" for v in top.values])
plt.axvline(0, color="black")
plt.xlabel("(test median - train median) / train std")
plt.title("Figure 1. Feature drift between train and test")
save_fig("14_feature_drift")
plt.show()
```

```python
biggest = shift.index[0]
print(f"Largest median shift: {biggest} ({shift[biggest]:+.2f} train std); "
      f"median {train[biggest].median():.3f} in train vs {test[biggest].median():.3f} in test.")
print(f"Features whose median moved by more than 0.2 train std: {shift[shift.abs() > 0.2].index.tolist() or 'none'}")

# Does the MEAN tell a different story? Compare the two measures feature by feature.
disagree = (mean_shift.abs() - shift.abs()).sort_values(ascending=False)
worst = disagree.index[0]
print()
print(f"Largest gap between the mean-based and median-based shift: {worst} "
      f"(mean {mean_shift[worst]:+.2f} vs median {shift[worst]:+.2f} train std).")
lo, hi = train[worst].min(), train[worst].max()
outside = ((test[worst] < lo) | (test[worst] > hi)).sum()
print(f"Training range of {worst}: {lo:.3f} to {hi:.3f}. Test rows outside that range: {outside:,}; "
      f"largest test value: {test[worst].max():,.1f}.")
print("Action -> the mean was misled by a handful of extreme rows, not by a general move of applicants. "
      "A production service should validate inputs (flag applications outside the training range) before scoring; "
      "and the cleaning rule for annual_inc in notebook 02 only catches values <= 0, so it should be tightened in a future version.")
```

### 11.4 Monitoring 2 - does performance hold up over time?
For every quarter of the test period we compute (a) the model's ROC-AUC, (b) the sub-grade's ROC-AUC (B1) on the same loans, and (c) the gap between the average predicted probability and the actual default rate.

**Figure 2.** Left: ROC-AUC by quarter, model vs B1. Right: predicted minus actual default rate by quarter.

```python
monitor = test[["issue_d", "default", "sub_grade_num"]].copy()
monitor["pred"] = test_scores
monitor["quarter"] = monitor["issue_d"].dt.to_period("Q").astype(str)

by_quarter = monitor.groupby("quarter").apply(lambda g: pd.Series({
    "loans": len(g),
    "model_auc": roc_auc_score(g["default"], g["pred"]),
    "b1_auc": roc_auc_score(g["default"], g["sub_grade_num"]),
    "mean_predicted": g["pred"].mean(),
    "actual_rate": g["default"].mean()}), include_groups=False)
by_quarter["lift_over_b1"] = by_quarter["model_auc"] - by_quarter["b1_auc"]
by_quarter["calibration_gap"] = by_quarter["mean_predicted"] - by_quarter["actual_rate"]
by_quarter["loans"] = by_quarter["loans"].astype(int)

fig, ax = plt.subplots(1, 2, figsize=(12, 4))
ax[0].plot(by_quarter.index, by_quarter["model_auc"], marker="o", label="model")
ax[0].plot(by_quarter.index, by_quarter["b1_auc"], marker="o", label="B1 sub_grade")
ax[0].set_ylabel("ROC-AUC"); ax[0].legend(); ax[0].tick_params(axis="x", rotation=60)
ax[1].bar(by_quarter.index, by_quarter["calibration_gap"], color="tab:purple")
ax[1].axhline(0, color="black"); ax[1].set_ylabel("Mean predicted - actual default rate")
ax[1].tick_params(axis="x", rotation=60)
fig.suptitle("Figure 2. Performance by quarter of the test period")
fig.tight_layout()
save_fig("15_performance_by_quarter")
plt.show()
by_quarter
```

```python
print(f"Model AUC by quarter ranges from {by_quarter['model_auc'].min():.3f} to {by_quarter['model_auc'].max():.3f}.")
print(f"Lift over B1 ranges from {by_quarter['lift_over_b1'].min():+.3f} to {by_quarter['lift_over_b1'].max():+.3f}; "
      f"it is positive in {(by_quarter['lift_over_b1'] > 0).sum()} of {len(by_quarter)} quarters.")
print(f"Calibration gap ranges from {by_quarter['calibration_gap'].min():+.3f} to {by_quarter['calibration_gap'].max():+.3f} "
      "(negative = the model predicts fewer defaults than happened).")
print(f"The model predicts fewer defaults than happened in {(by_quarter['calibration_gap'] < 0).sum()} of {len(by_quarter)} quarters.")
smallest = by_quarter["loans"].nsmallest(2)
print(f"The two quarters with the fewest loans are {smallest.index.tolist()} ({smallest.tolist()} loans); "
      "in notebook 02 (Table 4) the latest issue years are mostly unresolved, so their quarterly numbers are the least reliable.")
```

### 11.5 Retraining policy
A rule needs numbers. The two limits below are **policy choices we make here, not facts from the data**; a real lender would agree them with its risk team.

| Trigger | Limit (our assumption) |
|---|---|
| Ranking power is gone | Lift over B1 below **0.005** AUC for two quarters in a row |
| Probabilities are off | Absolute calibration gap above **0.05** in any quarter |

We apply the rule to the quarters we have, to see what it would have done.

```python
MIN_LIFT, MAX_GAP = 0.005, 0.05

low_lift = by_quarter["lift_over_b1"] < MIN_LIFT
by_quarter["trigger_lift"] = low_lift & low_lift.shift(1, fill_value=False)
by_quarter["trigger_gap"] = by_quarter["calibration_gap"].abs() > MAX_GAP
by_quarter["retrain"] = by_quarter["trigger_lift"] | by_quarter["trigger_gap"]

flagged = by_quarter[by_quarter["retrain"]]
print(by_quarter[["lift_over_b1", "calibration_gap", "trigger_lift", "trigger_gap", "retrain"]])
print()
print(f"The rule would have called for retraining in {len(flagged)} of {len(by_quarter)} quarters: "
      f"{flagged.index.tolist() or 'none'}.")
print(f"Quarters flagged only because of the calibration gap: {int((by_quarter['trigger_gap'] & ~by_quarter['trigger_lift']).sum())}; "
      "a persistent negative gap mostly reflects that the default rate itself is higher than in training, "
      "and the two last quarters are also the least mature.")
print("Action -> when triggered: re-fit the full pipeline on the most recent labelled loans, "
      "re-check against B1 on a later period, and only then replace the saved model.")
```

### 11.6 Model card
A model card records what the model is, what data it saw, and how well it did, so someone who finds `model.joblib` later knows what they have.

```python
card = {
    "model": config["model"],
    "task": "Predict probability that a Lending Club loan is charged off (1) vs fully paid (0)",
    "trained_on": {
        "rows": int(len(train)),
        "issue_dates": f"{train['issue_d'].min():%Y-%m-%d} to {train['issue_d'].max():%Y-%m-%d}",
        "default_rate": round(float(train["default"].mean()), 4)},
    "tested_on": {
        "rows": int(len(test)),
        "issue_dates": f"{test['issue_d'].min():%Y-%m-%d} to {test['issue_d'].max():%Y-%m-%d}",
        "default_rate": round(float(test["default"].mean()), 4)},
    "features": {"numeric": NUMERIC, "categorical": CATEGORICAL},
    "threshold": {"value": config["threshold"], "rule": config["rule"]},
    "test_metrics": {k: round(float(v), 4) for k, v in metrics.loc[config["model"]].items()},
    "baseline_B1_test_metrics": {k: round(float(v), 4) for k, v in metrics.loc["B1 sub_grade"].items()},
    "versions": {"python": platform.python_version(), "scikit-learn": sklearn.__version__,
                 "pandas": pd.__version__, "numpy": np.__version__},
    "known_limits": [
        "issue_d is used as a proxy for application date",
        "unfinished loans are excluded, so late-period labels are biased",
        "only accepted loans are in the data",
        "no fairness analysis (no protected attributes in the data)",
        "threshold maximises F1, not a real cost",
    ],
}
(ARTIFACTS / "model_card.json").write_text(json.dumps(card, indent=2))
print(json.dumps({k: card[k] for k in ["model", "trained_on", "tested_on", "threshold", "test_metrics"]}, indent=2))
```
